````python
import openai
import os
# from dotenv import load_dotenv, find_dotenv
# _ = load_dotenv(find_dotenv()) # read local .env file

# Ensure your API key is set as an environment variable or set it here
# openai.api_key ="value"

def get_completion(prompt, model="gpt-3.5-turbo"):
    
    messages = [{"role": "user", "content": prompt}]
    
    # Using the updated openai client library (as of late 2023/2024)
    try:
        client = openai.OpenAI()
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=0  # Use 0 for more predictable, factual output
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"An error occurred: {e}"

# Example Usage in a Jupyter Notebook Cell
prompt = """
Summarize the text delimited by triple backticks 
into a single sentence.

```The latest quarterly earnings report showed a 15% increase 
in revenue, primarily driven by the successful launch of 
the new 'Pro' model laptop. The stock price rose 5% in 
after-hours trading following the announcement.```
"""

response = get_completion(prompt)
print(response)
````
